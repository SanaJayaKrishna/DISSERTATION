```python
from pathlib import Path
import re
import pandas as pd
```

```python
tasks_dir = Path("./tasks")
records = []
```

```python
for task_file in sorted(tasks_dir.glob("*.task")):
    world_name = task_file.stem

    with open(task_file, "r", encoding="utf-8") as f:
        for line in f:
            line = line.strip()
            if not line:
                continue
            match = re.match(r"^(\d+)[\.\s]+(.*)$", line)
            if match:
                task_num = int(match.group(1))
                task_desc = match.group(2)
                records.append(
                    {"TaskNum": task_num, "World": world_name, "Task": task_desc}
                )

df = pd.DataFrame(records)
```
